```python
import pandas as pd
import matplotlib.pyplot as plt
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
# 1. Load the dataset
df = pd.read_csv("ecommerce_behavioral_dataset.csv")
```

```python
df.head()
```

```python
# 2. Preprocessing
df = df.drop_duplicates()
df = df.dropna()
```

```python
print("\nDataset after preprocessing:")
print(df.shape)
```

```python
# 3. Select and Purchase data
purchase_data = df[df["EventType"] == "Purchase"]
print("\nPurchase Data:")
print(purchase_data.head())
```

```python
# 4. Analyze customer preferences
popular_products = purchase_data["Product"].value_counts()
print("\nMost Popular Products:")
print(popular_products)
```

```python
# 5. Display graph
popular_products.head(10).plot(kind="bar", figsize=(8,5))

plt.title("Top 10 Purchased Products")
plt.xlabel("Product")
plt.ylabel("Number of Purchases")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 6. Create transaction basket
basket = pd.crosstab(
    purchase_data["TransactionID"],
    purchase_data["Product"]
)

# Convert to 0 and 1
basket = basket.applymap(lambda x: 1 if x > 0 else 0)

print("\nTransaction Basket:")
print(basket.head())
```

```python
# 7. Apply Apriori algorithm
frequent_items = apriori(
    basket,
    min_support=0.05,
    use_colnames=True
)

print("\nFrequent Itemsets:")
print(frequent_items)
```

```python
# 7. Apply Apriori algorithm
frequent_items = apriori(
    basket,
    min_support=0.05,
    use_colnames=True
)

print("\nFrequent Itemsets:")
print(frequent_items)
```

```python
# 8. Generate association rules
rules = association_rules(
    frequent_items,
    metric="confidence",
    min_threshold=0.4
)

# Keep strong rules
rules = rules[rules["lift"] > 1]

print("\nAssociation Rules:")

if not rules.empty:

    print(
        rules[
            ["antecedents",
             "consequents",
             "support",
             "confidence",
             "lift"]
        ].head(10)
        )

else:
    print("No strong rules found.")
```

```python
# 9. Product Recommendation
product = "Mobile Phone"

print("\nRecommendations for:", product)

if not rules.empty:

    recommendations = rules[
        rules["antecedents"].apply(
            lambda x: product in x
        )
    ].sort_values(
        "confidence",
        ascending=False
    )

    if not recommendations.empty:

        for _, row in recommendations.iterrows():

            for item in row["consequents"]:
                print("-", item)

    else:
        print("No recommendation found.")

else:
    print("No recommendation available.")
```
